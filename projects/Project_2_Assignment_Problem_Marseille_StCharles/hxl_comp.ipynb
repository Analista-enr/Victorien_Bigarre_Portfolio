```python
import pandas as pd
from data_process import TrainScheduler
from Hxl_model import HexalySolver
```

```python
CAMINO = r"C:\Users\Elfamoso\OneDrive\Bureau\Veille\Portfolio\Tchoutchou\Planning_original.xlsx"


Data = TrainScheduler(pd.read_excel(CAMINO), nb_quai=16, penalty=100)
df = Data.arrivals
df = df.fillna(0)
```

```python
nb_quais = 16  # Nombre de quais disponibles
```

```python
# Créer et résoudre le modèle
model = HexalySolver(df, nb_quais, penalty=10)
model.solve(time_limit=120)  # 2 minutes max
```

```python
# Récupérer la solution optimale
N_optimal = (
    model.get_solution()
)  # Liste des affectations [quai_train_0, quai_train_1, ...]
score_optimal = model.get_objective_value()  # Valeur de op_score minimale
```

```python
print(f"Solution optimale: {N_optimal}")
print(f"Score minimal: {score_optimal}")
print(f"Statut: {model.get_status()}")
```
